## 3. Listas, dicionários e estruturas de controle

Listas permitem armazenar vários valores em uma mesma variável.

````python
notas = [7.5, 8.0, 6.5, 9.0, 10.0]

print("Lista completa:", notas)
print("Primeira nota:", notas[0])
print("Última nota:", notas[-1])
print("Quantidade de notas:", len(notas))
````

Um **dicionário** armazena informações no formato `chave: valor`.

````python
aluno1 = {
    "nome": "Carlos",
    "idade": 22,
    "curso": "Engenharia de Computação",
    "nota": 8.5
}

aluno2 = {
    "nome": "José",
    "idade": 23,
    "curso": "Engenharia de Computação",
    "nota": 9.5
}

print(aluno1)
print("Nome:", aluno1["nome"])
print("Nota:", aluno1["nota"])

print(aluno2)
print("Nome:", aluno2["nome"])
print("Nota:", aluno2["nota"])
````

### Condições

````python
nota = 7.5

if nota >= 7:
    print("Aprovado")
elif nota >= 5:
    print("Recuperação")
else:
    print("Reprovado")
````

## 4. NumPy: trabalhando com dados numéricos

O **NumPy** é uma biblioteca fundamental para computação numérica em Python.

Por convenção, importamos a biblioteca como `np`.

````python
import numpy as np

valores = np.array([10, 12, 15, 18, 20])

print(valores)
print("Média:", np.mean(valores))
print("Mediana:", np.median(valores))
print("Variância:", np.var(valores))
print("Desvio-padrão:", np.std(valores))
print("Mínimo:", np.min(valores))
print("Máximo:", np.max(valores))
````

### Média, variância e desvio-padrão

A **média** indica uma tendência central dos valores.

A **variância** mede o quanto os valores se espalham em relação à média. Quanto maior a variância, maior tende a ser a dispersão.

O **desvio-padrão** é a raiz quadrada da variância e fica na mesma unidade dos dados originais.

## 6. Carregando um dataset do scikit-learn

O **scikit-learn** possui diversos datasets que podem ser utilizados para aprendizado e experimentação.

Nesta aula utilizaremos o **Iris Dataset**.

Ele possui medidas de flores de três espécies de íris:

- setosa;
- versicolor;
- virginica.

As características são:

- comprimento da sépala;
- largura da sépala;
- comprimento da pétala;
- largura da pétala.

O objetivo clássico de Machine Learning é aprender a identificar a espécie da flor a partir dessas medidas.

````python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)

print(type(iris))
print(iris.keys())
````

### Transformando os dados em um DataFrame

````python
df = iris.frame.copy()

df.head()
````

O dataset utiliza inicialmente números (`0`, `1` e `2`) para representar as espécies. Vamos criar uma coluna com os nomes.

````python
iris.target_names
````

````python
iris.target
````

````python
df["species"] = df["target"].map(
    dict(enumerate(iris.target_names))
)

df.head()
````

## 7. Explorando o dataset

Antes de aplicar qualquer algoritmo de Inteligência Artificial, é importante **conhecer os dados**.

````python
print("Número de linhas e colunas:", df.shape)

print("\nNomes das colunas:")
print(df.columns.tolist())

print("\nTipos dos dados:")
print(df.dtypes)
````

### Informações gerais

````python
df.info()
````

### Verificando valores ausentes

Valores ausentes podem causar problemas em diversos algoritmos.

````python
df.isnull().sum()
````

### Quantidade de exemplos de cada espécie

````python
df["species"].value_counts()
````

## 8. Estatística descritiva

O método `describe()` calcula várias estatísticas automaticamente.

````python
df.describe()
````

Agora vamos calcular algumas medidas individualmente para o **comprimento da pétala**.

````python
coluna = "petal length (cm)"

print("Média:", df[coluna].mean())
print("Mediana:", df[coluna].median())
print("Variância:", df[coluna].var())
print("Desvio-padrão:", df[coluna].std())
print("Mínimo:", df[coluna].min())
print("Máximo:", df[coluna].max())

print("\nQuartis:")
print(df[coluna].quantile([0.25, 0.50, 0.75]))
````

### Estatísticas por espécie

O `groupby()` permite agrupar os dados de acordo com alguma característica.

````python
media_por_especie = df.groupby("species").mean(numeric_only=True)

media_por_especie
````

Observe as diferenças entre as médias das três espécies. Elas já sugerem que algumas características podem ser úteis para separá-las.

````python
df.groupby("species")["petal length (cm)"].agg(
    ["mean", "median", "var", "std", "min", "max"]
)
````

## 9. Visualização dos dados

Gráficos ajudam a identificar padrões, diferenças, dispersão e possíveis valores atípicos.

Vamos utilizar o **Matplotlib**.

````python
import matplotlib.pyplot as plt
````

### Histograma

O histograma mostra como os valores de uma variável estão distribuídos.

````python
plt.figure(figsize=(8, 5))

plt.hist(df["sepal length (cm)"], bins=12)

plt.xlabel("Comprimento da sépala (cm)")
plt.ylabel("Frequência")
plt.title("Distribuição do comprimento da sépala")

plt.show()
````

## 10. Boxplot

O **boxplot** é muito utilizado para visualizar a distribuição dos dados.

Ele permite observar:

- mediana;
- primeiro quartil (Q1);
- terceiro quartil (Q3);
- dispersão;
- possíveis valores atípicos (*outliers*).

A caixa contém aproximadamente os 50% centrais dos dados.

````python
plt.figure(figsize=(7, 5))

plt.boxplot(df["sepal length (cm)"])

plt.ylabel("Comprimento da sépala (cm)")
plt.title("Boxplot do comprimento da sépala")

plt.show()
````

### Boxplot comparando as espécies

Agora conseguimos observar se a distribuição de uma variável muda de acordo com a espécie.

````python
df.boxplot(
    column="petal length (cm)",
    by="species",
    figsize=(8, 5)
)

plt.xlabel("Espécie")
plt.ylabel("Comprimento da pétala (cm)")
plt.title("Comprimento da pétala por espécie")
plt.suptitle("")

plt.show()
````

## 11. Gráfico de dispersão

Um gráfico de dispersão permite analisar a relação entre **duas variáveis**.

````python
plt.figure(figsize=(8, 6))

for especie in df["species"].unique():
    dados_especie = df[df["species"] == especie]

    plt.scatter(
        dados_especie["petal length (cm)"],
        dados_especie["petal width (cm)"],
        label=especie
    )

plt.xlabel("Comprimento da pétala (cm)")
plt.ylabel("Largura da pétala (cm)")
plt.title("Comprimento × largura da pétala")
plt.legend()

plt.show()
````

## 12. Correlação entre variáveis

A correlação ajuda a identificar se duas variáveis tendem a variar juntas.

Valores próximos de:

- **+1:** forte relação positiva;
- **0:** pouca relação linear;
- **-1:** forte relação negativa.

````python
colunas_numericas = iris.feature_names

correlacao = df[colunas_numericas].corr()

correlacao
````

Vamos visualizar essa matriz de forma simples.

````python
plt.figure(figsize=(8, 6))

imagem = plt.imshow(correlacao, aspect="auto")
plt.colorbar(imagem)

plt.xticks(
    range(len(colunas_numericas)),
    colunas_numericas,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(colunas_numericas)),
    colunas_numericas
)

plt.title("Matriz de correlação")

plt.tight_layout()
plt.show()
````

## 13. Seleção e criação de novos dados

## 14. Um primeiro contato com treino e teste

Em Machine Learning, normalmente não utilizamos todos os dados para treinar um modelo.

Uma divisão muito comum é:

- **conjunto de treinamento:** utilizado pelo algoritmo para aprender;
- **conjunto de teste:** utilizado para avaliar o modelo em exemplos que ele não viu durante o treinamento.

Neste momento vamos apenas fazer a divisão, sem treinar nenhum algoritmo.

````python
from sklearn.model_selection import train_test_split

X = df[iris.feature_names]
y = df["target"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Dataset completo:", X.shape)
print("Treinamento:", X_train.shape)
print("Teste:", X_test.shape)
````
